### Task 2: Dimensionality Reduction with PCA:

#### Standardization

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
x=data.drop("Type",axis=1)
y=data["Type"]
```

```python
scaled_data=scaler.fit_transform(x)
scaled_data
```

```python
sc_df=pd.DataFrame(scaled_data,columns=x.columns)
sc_df.head()
```

#### PCA 

```python
from sklearn.decomposition import PCA
pca=PCA()
pca_data=pca.fit_transform(scaled_data)
pca_data
```

#### Explained Variance

```python
explained_var=pca.explained_variance_ratio_
explained_var
```

```python
import numpy as np
```

### Cumulative Explained Variance

```python
cum_var=np.cumsum(explained_var)
cum_var
```

#### Scree plot

```python
plt.plot(range(1,len(explained_var)+1),explained_var,marker="o")
plt.title("Scree Plot",color="red")
plt.xlabel("Principal Component")
plt.ylabel("Explained Variance Ratio")
plt.show()
```

#### Cumulative Explained Variance Plot

```python
plt.figure(figsize=(10,6))
plt.plot(range(1,len(cum_var)+1),cum_var,marker="o")
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Cumulative Explained Variance",color="red")
plt.show()
```

```python
n_components=np.argmax(cum_var>=0.95)+1
print("n_components=",n_components)
```

```python
pca=PCA(n_components=n_components)
x_pca=pca.fit_transform(scaled_data)
x_pca
```

### K Means

```python
from sklearn.cluster import KMeans
import os
os.environ["OMP_NUM_THREADS"] = "1"
```

```python
from sklearn.metrics import silhouette_score
from sklearn.metrics import davies_bouldin_score
```

```python
k_values=range(2,9)
original_results=[]
for i in k_values:
    kmeans=KMeans(n_clusters=i,random_state=42,n_init=10)
    labels=kmeans.fit_predict(scaled_data)
    silhouette=silhouette_score(scaled_data,labels)
    db_index=davies_bouldin_score(scaled_data,labels)
    original_results.append({"K":i,"Silhouette Score":silhouette,"Davies-Bouldin Index":db_index})
og_r=pd.DataFrame(original_results)
og_r
```

#### Plot silhouette score

#### Conclusion

The Wine dataset contains 13 numerical features and was first explored using descriptive statistics, histograms, box plots, density plots, and correlation analysis. The features were standardized before applying PCA because they have different scales. PCA reduced the 13 original features to 10 principal components while retaining approximately 96.17% of the total variance.

K-Means clustering was performed on both the original standardized data and the PCA-transformed data. For both approaches, K=3 produced the highest Silhouette Score among the tested values from 2 to 8. The original standardized data achieved a Silhouette Score of approximately 0.2849 and a Davies-Bouldin Index of approximately 1.3892. The PCA data achieved a slightly higher Silhouette Score of approximately 0.2987 and a lower Davies-Bouldin Index of approximately 1.3363.

These results show that PCA reduced the dimensionality from 13 to 10 features while retaining most of the information and produced slightly better clustering metrics for this dataset. PCA is useful when dimensionality reduction, visualization, and reduction of redundant information are important. Clustering on the original standardized features is useful when maintaining the direct interpretability of the original variables is important.
